 2. Basic Try-Except Block

```python


try:
    x = int(input("Enter a number: "))
    print("You entered:", x)
except ValueError:
    print("That was not a valid number!")
```

3. Try-Except-Else-Finally Structure

```python
try:
    num = int(input("Enter a number to divide 100: "))
    result = 100 / num
except ZeroDivisionError:
    print("Cannot divide by zero!")
except ValueError:
    print("Invalid input! Not a number.")
else:
    print("Result is:", result)
finally:
    print("Execution complete. This block always runs.")
```

4. Catching Multiple Exceptions

```python
try:
    a = int("ten")
    b = 10 / 0
except (ValueError, ZeroDivisionError) as e:
    print("An error occurred:", e)
```

 5. Raising Custom Exceptions

```python


def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    else:
        print("Age is valid.")

try:
    check_age(-5)
except ValueError as e:
    print("Caught an exception:", e)
```

 6. DEBUGGING PRACTICES

 A. Using print() statements

```python
def divide(a, b):
    print(f"Dividing {a} by {b}")
    return a / b

print(divide(10, 2))

```

B. Using traceback module (for detailed errors)

```python
import traceback

try:
    result = 10 / 0
except Exception as e:
    print("Something went wrong.")
    traceback.print_exc()
```

 C. Using the Python Debugger (pdb)

```python


def calculate_total(price, quantity):

    return price * quantity

print(calculate_total(50, 2))
```
