Writing and Using Functions in Python


Defining and calling functions

Positional and default arguments

*args (variable arguments)

**kwargs (keyword arguments)



```python


def greet():
    print("Hello! Welcome to Python Functions.")

greet()
```

```python
FUNCTION WITH PARAMETERS

def greet_user(name):
    print(f"Hello, {name}!")

greet_user("Alice")
greet_user("Bob")
```

```python
 FUNCTION WITH DEFAULT PARAMETERS

def greet_user(name="Guest"):
    print(f"Welcome, {name}!")

greet_user()
greet_user("Ravi")
```

```python

def print_names(*names):
    print("Names passed:", names)
    for name in names:
        print("Hello", name)

print_names("Alice", "bob", "Charlie")
```

```python


def print_info(**info):
    print("User Info:")
    for key, value in info.items():
        print(f"{key} = {value}")

print_info(name="Alice", age=25, city="Delhi")
```

```python
COMBINING *args AND **kwargs

def show_details(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

show_details(1, 2, 3, name="Ravi", city="Mumbai")
```

```python
 FUNCTION NESTING (CALL ONE FUNCTION INSIDE ANOTHER)

def multiply(a, b):
    return a * b

def square(n):
    return multiply(n, n)

print("Square of 4:", square(4))
```
