Task 1 — Analysis Using Python Basics
Store the given sensor readings using appropriate Python data structures, then perform the following tasks.

2. Identify high temperature
Apply the following condition:

Display: “High Temperature Alert”
Otherwise, display: “Temperature Normal”

3. Identify low battery
Apply the following condition:

1.

```python
# Battery sensor data
battery_data = [
    {"Time": 0, "Temperature": 32, "Voltage": 400, "Current": 20, "Charge": 95},
    {"Time": 1, "Temperature": 35, "Voltage": 395, "Current": 22, "Charge": 88},
    {"Time": 2, "Temperature": 38, "Voltage": 390, "Current": 25, "Charge": 80},
    {"Time": 3, "Temperature": 40, "Voltage": 385, "Current": 28, "Charge": 70},
    {"Time": 4, "Temperature": 42, "Voltage": 380, "Current": 30, "Charge": 60},
    {"Time": 5, "Temperature": 45, "Voltage": 375, "Current": 32, "Charge": 48}
]

# Display all battery readings using a for loop
for reading in battery_data:
    print(
        f"Time: {reading['Time']} h, "
        f"Temperature: {reading['Temperature']} °C, "
        f"Voltage: {reading['Voltage']} V, "
        f"Current: {reading['Current']} A, "
        f"Charge: {reading['Charge']}%"
    )
```

2.

```python
# Check temperature condition
for reading in battery_data:
    temperature = reading["Temperature"]

    if temperature > 40:
        print(f"Time {reading['Time']} h: High Temperature Alert")
    else:
        print(f"Time {reading['Time']} h: Temperature Normal")
```

3.

```python
# Check battery charge condition
for reading in battery_data:
    charge = reading["Charge"]

    if charge < 50:
        print(f"Time {reading['Time']} h: Low Battery Alert")
    else:
        print(f"Time {reading['Time']} h: Battery Charge Normal")
```

4.

```python
def battery_status(charge):
    if charge > 70:
        return "High"
    elif 50 <= charge <= 70:
        return "Medium"
    else:
        return "Low"


# Test the function for each reading
for reading in battery_data:
    status = battery_status(reading["Charge"])
    print(f"Time {reading['Time']} h: Charge = {reading['Charge']}%, Status = {status}")
```

Task 4 — Analysis Using Calculus
The battery charge is modeled as a function of time:

C(t) = 100 − 2t − ½t²
where C(t) is the battery charge percentage and t is the time in hours.
1. Find the rate of change of battery charge
The rate at which the battery charge changes is given by dC/dt. Differentiate C(t) and determine dC/dt.
2. Calculate the discharge rate
Find the discharge rate at t = 2 and t = 5, that is, calculate:

dC/dt at t = 2 and dC/dt at t = 5

3. Interpret the result
Based on the values of dC/dt, determine whether the battery is losing charge:
● At a constant rate
● At an increasing rate
● At a decreasing rate


1.

```python
import sympy as sp

t = sp.symbols('t')

C = 100 - 2*t - sp.Rational(1, 2)*t**2

# Differentiate C(t)
dC_dt = sp.diff(C, t)

print("C(t) =", C)
print("dC/dt =", dC_dt)
```

2.

```python
# Calculate discharge rates
rate_t2 = dC_dt.subs(t, 2)
rate_t5 = dC_dt.subs(t, 5)

print("dC/dt at t = 2:", rate_t2, "% per hour")
print("dC/dt at t = 5:", rate_t5, "% per hour")
```
